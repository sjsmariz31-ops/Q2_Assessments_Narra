# MODULAR CALCULATOR

1.

Operation	- addition

Function Name -	add_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 + num2

Return Value - sum of two numbers


2.

Operation	- subtraction

Function Name -	subtract_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 - num2

Return Value - difference of two numbers

3.

Operation	- multiplication

Function Name -	multiply_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 * num2

Return Value - product of two numbers

4.

Operation	- division

Function Name -	divide_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 / num2

Return Value - quotient of two numbers

```python




num1 = float(input("enter first number:"))
num2 = float(input("enter second number:"))

operation = float(input("1 - Addition 2 - Subtraction 3 - Multiplication 4 - Division:"))


def add_numbers (num1, num2):
    sum = num1 + num2
    return sum

def subtract_numbers (num1, num2):
    Difference = num1 - num2
    return Difference

def multiply_numbers (num1, num2):
    product = num1 * num2
    return product

def divide_numbers (num1, num2):
    quotient = num1 / num2
    return quotient

if operation == 1 :
   print(add_numbers(num1,num2))

elif operation == 2 :
  print(subtract_numbers(num1, num2))

elif operation == 3 :
  print(multiply_numbers(num1, num2))

elif operation == 4 :
  print(divide_numbers(num1, num2))
```
