## MODULAR CALCULATOR

1.

Operation	- addition

Function Name -	add_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 + num2

Return Value - sum of two numbers


2.

Operation	- subtraction

Function Name -	subtract_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 - num2

Return Value - difference of two numbers

3.

Operation	- multiplication

Function Name -	multiply_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 * num2

Return Value - product of two numbers

4.

Operation	- division

Function Name -	divide_numbers

Parameters Needed	- num1, num2

Process Inside the Function	 - num1 / num2

Return Value - quotient of two numbers

```python
# Ask the user for two numbers
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Ask the user to choose an operation
print("Choose operation:")
print("1 - Addition")
print("2 - Subtraction")
print("3 - Multiplication")
print("4 - Division")
operation = int(input("Enter choice (1/2/3/4):"))

# Function for addition
def add_numbers(num1, num2):
    sum_result = num1 + num2
    return sum_result

# Function for subtraction
def subtract_numbers(num1, num2):
    difference = num1 - num2
    return difference

# Function for multiplication
def multiply_numbers(num1, num2):
    product = num1 * num2
    return product

# Function for division with
def divide_numbers(num1, num2):
    if num2 == 0:
        return "Cannot divide by zero"
    quotient = num1 / num2
    return quotient

# Call the correct values
if operation == 1:
    result = add_numbers(num1, num2)
    print("Result:", result)
elif operation == 2:
    result = subtract_numbers(num1, num2)
    print("Result:", result)
elif operation == 3:
    result = multiply_numbers(num1, num2)
    print("Result:", result)
elif operation == 4:
    result = divide_numbers(num1, num2)
    print("Result:", result)
else:
    print("Invalid operation choice")
```
